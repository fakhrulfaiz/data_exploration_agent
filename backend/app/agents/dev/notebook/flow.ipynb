```python
from agent.main_agent import build_main_agent

from langgraph.checkpoint.memory import MemorySaver

checkpointer = MemorySaver()
config = {"configurable": {"thread_id": "dev-fyp"}}

main_agent = build_main_agent(checkpointer=checkpointer)
```

```python
# Example usage
questions = [
    # "Which genre has the oldest painting?",\
    # "Does the oldest painting has one person in it?",
    "Get the number of paintings that shows Fruit for each century.",
]
```

```python
for question in questions:
    print(f"\nQuestion: {question}")
    print("-" * 50)
    
    for step in main_agent.stream(
        {"messages": [{"role": "user", "content": question}]},
        stream_mode="values",
        config=config,
    ):
        step["messages"][-1].pretty_print()
```

```python
from langgraph.types import Command
interrupts = []
for step in main_agent.stream(
    Command(resume=True), 
    stream_mode="values",
    config=config,
    checkpointer=checkpointer,
):
    step["messages"][-1].pretty_print() 
```
